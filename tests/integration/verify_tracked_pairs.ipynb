```python
import sys
from pathlib import Path
import pandas as pd

# Add src to path
src_path = Path().absolute().parent.parent / 'src'
sys.path.insert(0, str(src_path))

from cryptoquant.database.session import get_session
from cryptoquant.database.models import TrackedPair
```

```python
# Test database connection
session = get_session()
print("✅ Database connection successful")
```

```python
# Query tracked pairs
tracked_pairs = session.query(TrackedPair).all()
print(f"Found {len(tracked_pairs)} tracked pairs:\n")

for pair in tracked_pairs:
    status = "✓ Active" if pair.is_tracking_active else "✗ Inactive"
    print(f"{status} | {pair.product_id} ({pair.symbol})")
    print(f"   Created: {pair.created_at}")
```

```python
# Display as DataFrame
df = pd.DataFrame([
    {
        'id': p.id,
        'product_id': p.product_id,
        'symbol': p.symbol,
        'is_tracking_active': p.is_tracking_active,
        'created_at': p.created_at
    }
    for p in tracked_pairs
])
df
```

```python
session.close()
```
